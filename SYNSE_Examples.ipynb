This notebook demonstrates how to use the SYNSE toolkit for creating, parsing, visualizing, and working with SysML v2 models.

## 1. Installation and Setup

Make sure you have installed the SYNSE package with all dependencies.

````python
# Verify installation
from sysml_parser import Model, Class, Package, AttributeUsage
print("✓ SYNSE toolkit imported successfully!")
````

## 2. Creating Your First Model

Let's create a simple automotive system model with packages, classes, and attributes.

````python
# Create a model
model = Model(
    identifier="automotive_system",
    name="Automotive System",
    author="Your Name",
    description="A sample automotive system model"
)

# Create a package
powertrain_pkg = Package(
    identifier="powertrain_pkg",
    name="Powertrain Package",
    version="1.0"
)
model.add_package(powertrain_pkg)

# Create a class
engine = Class(
    identifier="Engine",
    name="Engine",
    documentation="Internal combustion engine"
)
powertrain_pkg.add_member(engine)

# Add attributes
displacement = AttributeUsage(
    identifier="displacement",
    name="displacement",
    feature_type="Real",
    documentation="Engine displacement in cc"
)
engine.add_feature(displacement)

power = AttributeUsage(
    identifier="power",
    name="power",
    feature_type="Real",
    documentation="Engine power in kW"
)
engine.add_feature(power)

print(f"✓ Model created: {model.name}")
print(f"  Package: {powertrain_pkg.name}")
print(f"  Class: {engine.name}")
#print(engine.features)
print(f"  Features: {[f for f in engine.features]}")
````

## 3. Writing Models to Files

Export the model to JSON format.

````python
from sysml_parser import SysMLWriter

# Save the model
writer = SysMLWriter(model)
writer.write_file("automotive_model.json", format="json")

print("✓ Model saved to automotive_model.json")

# Get JSON string representation
json_str = writer.to_json_string()
print(f"\nJSON Output (first 500 chars):\n{json_str[:500]}...")
````

## 4. Parsing and Loading Models

Load and parse a SysML model from a file.

````python
from sysml_parser import SysMLParser

# Parse the model
parser = SysMLParser()
loaded_model = parser.parse_file("automotive_model.json")

# Access elements
print(f"Model: {loaded_model.name}")
print(f"Author: {loaded_model.author}")
print(f"Description: {loaded_model.description}")

# Explore structure
print("\nStructure:")
for pkg_id, package in loaded_model.owned_packages.items():
    print(f"  Package: {package.name}")
    for member_id, member in package.members.items():
        print(f"    └─ {member.name} ({member.__class__.__name__})")
        if hasattr(member, 'features') and member.features:
            for feature in member.features:
                print(f"       └─ {feature.name} ({feature.feature_type})")
````

## 5. Visualizing Models

Generate various visualizations of your model.

````python
from sysml_parser.views import ModelViewer

# Create a viewer
viewer = ModelViewer(model)

# Generate model statistics
print("=== Model Statistics ===")
print(viewer.view_statistics())
````

````python
# Generate hierarchical tree view
print("\n=== Model Tree View ===")
print(viewer.view_tree(max_depth=3))
````

````python
# Generate features view
print("\n=== Model Features ===")
print(viewer.view_features())
````

## 6. Creating Complex Models with Functions and Ports

Build more advanced models with functions, ports, and requirements.

````python
from sysml_parser import (
    Function, PortUsage, PartUsage, Requirement, 
    Multiplicity, Constraint
)

# Create a new model
advanced_model = Model(
    identifier="aircraft_system",
    name="Aircraft Control System",
    author="Systems Engineer",
    description="High-level aircraft control system"
)

# Create packages
avionics_pkg = Package(identifier="avionics", name="Avionics")
advanced_model.add_package(avionics_pkg)

# Create autopilot class
autopilot = Class(
    identifier="Autopilot",
    name="Autopilot System",
    documentation="Automated flight control"
)
avionics_pkg.add_member(autopilot)

# Add sensor ports (incoming)
altitude_sensor = PortUsage(
    identifier="altitude_in",
    name="altitude_sensor",
    feature_type="AltitudeSensor",
    port_direction="incoming"
)
autopilot.add_feature(altitude_sensor)

# Add actuator ports (outgoing)
control_surface = PortUsage(
    identifier="control_out",
    name="control_surface",
    feature_type="ControlActuator",
    port_direction="outgoing"
)
autopilot.add_feature(control_surface)

# Add internal parts
processor = PartUsage(
    identifier="cpu",
    name="Flight Computer",
    feature_type="Processor",
    multiplicity=Multiplicity(1, 1)
)
autopilot.add_feature(processor)

# Create function
control_func = Function(
    identifier="control_logic",
    name="Flight Control Logic",
    documentation="Main autopilot control algorithm"
)
avionics_pkg.add_member(control_func)

# Add function I/O
altitude_in = AttributeUsage(
    identifier="altitude",
    name="altitude",
    feature_type="Real"
)
control_func.add_input(altitude_in)

control_out = AttributeUsage(
    identifier="elevator_cmd",
    name="elevator_command",
    feature_type="Real"
)
control_func.add_output(control_out)

print("✓ Complex model created successfully!")
````

````python
# Add requirements to the model
req1 = Requirement(
    identifier="REQ_001",
    name="Altitude Control",
    requirement_id="REQ-001",
    text="The autopilot shall control altitude within ±100 feet",
    satisfied_by=["control_logic"]
)
avionics_pkg.add_member(req1)

# Add constraints
constraint = Constraint(
    identifier="constraint_1",
    name="Control Response Time",
    expression="response_time < 100ms",
    constrained_elements=["Autopilot"]
)
avionics_pkg.add_member(constraint)

print("✓ Requirements and constraints added!")

# Visualize the advanced model
viewer2 = ModelViewer(advanced_model)
print("\n=== Advanced Model Statistics ===")
print(viewer2.view_statistics())
````

## 7. Working with Multiplicity

Define and use multiplicity constraints for element cardinality.

````python
from sysml_parser import Multiplicity

# Create various multiplicity constraints
multiplicities = [
    (Multiplicity(0, 1), "Optional (0..1)"),
    (Multiplicity(1, 1), "Exactly one (1)"),
    (Multiplicity(0, -1), "Many (*)"),
    (Multiplicity(1, -1), "One or more (1..*)")
]

print("Multiplicity Examples:")
for mult, description in multiplicities:
    print(f"  {str(mult):10} -> {description}")
````

## 8. Reading and Processing Model Elements

Query and manipulate model elements programmatically.

````python
# Get specific element
element = parser.get_element("Engine")
if element:
    print(f"Found element: {element.name}")
    print(f"  Type: {element.__class__.__name__}")
    print(f"  Documentation: {element.documentation}")

# Get elements by type
classes = parser.get_elements_by_type(Class)
print(f"\nFound {len(classes)} class(es):")
for cls in classes:
    print(f"  - {cls.name} (ID: {cls.identifier})")
````
